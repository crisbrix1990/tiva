```python
# IMPORTS NECESARIOS
# ===============================
import os
import glob
import pandas as pd
from pathlib import Path
#from google.colab import drive
```

```python
# ===============================
# DEFINIR DIRECTORIOS
# ===============================
base_dir = Path(r'G:\Mi unidad\Proyecto TIVA\tiva')
print(f"Directorio base: {base_dir}")

ruta = base_dir / 'Bases_95_2020_csv'
ruta_pickle = base_dir / 'Bases_952020_pickles'

# ===============================
# BUSCAR ARCHIVOS CSV
# ===============================
patron = str(ruta / "*[0-9][0-9][0-9][0-9]_SML.csv")
csv_files = glob.glob(patron)

print(f"Archivos encontrados: {len(csv_files)}")

# ===============================
# COLUMNAS A ELIMINAR
# ===============================
columns_to_drop_patterns = [
    '_INVNT', '_DPABR', '_HFCE',
    '_NPISH', '_GGFC', '_GFCF'
]

# ===============================
# PROCESAMIENTO
# ===============================
for file in csv_files:
    df = pd.read_csv(file)

    # Eliminar columnas usando regex
    regex_pattern = '|'.join(columns_to_drop_patterns)
    columns_to_drop = df.filter(regex=regex_pattern).columns
    df = df.drop(columns_to_drop, axis=1)

    # Extraer año del nombre del archivo
    año = Path(file).stem.split('_')[0]

    # Agregar columna Año
    df['Año'] = int(año)

    # Guardar pickle
    pickle_file = ruta_pickle / f"{año}_SML.pkl"
    df.to_pickle(pickle_file)

    print(f"Archivo guardado: {pickle_file}")
```

```python
# ===============================
# BUSCAR ARCHIVOS PICKLE
# ===============================
pickle_files = list(ruta_pickle.glob("*_SML.pkl"))

print(f"Archivos pickle encontrados: {len(pickle_files)}")

if len(pickle_files) == 0:
    raise FileNotFoundError("No se encontraron archivos pickle en la ruta indicada.")

# ===============================
# CARGAR DATAFRAMES
# ===============================
dfs = []

for file_path in pickle_files:
    try:
        df = pd.read_pickle(file_path)
        dfs.append(df)
        print(f"Cargado: {file_path.name}")
    except Exception as e:
        print(f"Error al cargar {file_path.name}: {e}")

# ===============================
# CONCATENAR
# ===============================
combined_df = pd.concat(dfs, ignore_index=True)

# ===============================
# GUARDAR DATASET FINAL
# ===============================
final_pickle_file = ruta_pickle / "combined_data.pkl"
combined_df.to_pickle(final_pickle_file)

print(f"DataFrame combinado guardado en: {final_pickle_file}")
print(f"Filas totales: {combined_df.shape[0]}")
print(f"Columnas totales: {combined_df.shape[1]}")
```
